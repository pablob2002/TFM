# Active users analysis with links.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
```

```python
# Load the dataset
# import datasets
df = pd.read_json("../../../Datasets/voluntariosdanavalencia_old.json") # Import data from Voluntarios de Valencia groupchat.
df_links = pd.read_csv("../../../Datasets/voluntariosdanavalencia_extracted_links.csv") # Import data from Voluntarios de Valencia groupchat with messages with links and misinformation score.
df_users = pd.read_csv("../../../Datasets/voluntariosdanavalencia_extracted_users.csv") # Import data from Voluntarios de Valencia groupchat with user information.
```

```python
# Define a function to extract messages with links 
def contains_link(text):
    """Return True if the text contains a URL."""
    if not isinstance(text, str):
        return False
    # Simple regex to match common URL patterns
    url_pattern = re.compile(
        r'http[s]?://(?:[a-zA-Z]|[0-9]|[$-_@.&+]|[!*\\(\\),]|(?:%[0-9a-fA-F][0-9a-fA-F]))+'
        r'|www\.[a-zA-Z0-9\-]+(?:\.[a-zA-Z]{2,})+'
    )
    return bool(url_pattern.search(text))
```

## Check number of messages with links per channel

```python
# Check number of messages containing links per channel
df['has_link'] = df['message'].apply(contains_link)
link_counts = df.groupby('topic_name')['has_link'].sum().reset_index(name='link_count')
# Display the link counts per channel
print(link_counts)
```

```python
# Percentage of messages containing links per channel and number of messages per channel
link_counts_current = (
    df.groupby('topic_name')['has_link']
    .sum()
    .reset_index(name='link_count')
)
message_counts = (
    df.groupby('topic_name')['message']
    .count()
    .reset_index(name='message_count')
)

channel_link_stats = link_counts_current.merge(
    message_counts,
    on='topic_name',
    how='inner',
    validate='one_to_one'
 )
channel_link_stats['link_percentage'] = (
    channel_link_stats['link_count'] / channel_link_stats['message_count']
 ) * 100
channel_link_stats['total_messages'] = channel_link_stats['message_count']

print(channel_link_stats[['topic_name', 'link_percentage', 'total_messages']])
```

```python
# Sort channels by link percentage and visualize
channel_link_stats_sorted = channel_link_stats.sort_values(
    by=['link_percentage', 'total_messages'],
    ascending=[False, False]
).reset_index(drop=True)

print(channel_link_stats_sorted[['topic_name', 'link_percentage', 'total_messages']])

plt.figure(figsize=(12, 6))
plt.bar(
    channel_link_stats_sorted['topic_name'],
    channel_link_stats_sorted['link_percentage'],
    color='teal',
    alpha=0.85
)
plt.title('Percentage of Messages with Links by Channel')
plt.xlabel('Channel')
plt.ylabel('Link percentage (%)')
plt.xticks(rotation=75, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Top 10 most active users sending messages with links
top_link_senders = (
    df_links['sender_id']
    .value_counts()
    .head(10)
)
print("Top 10 most active users sending messages with links:")
print(top_link_senders) 

# Total number of messages with links
total_links = df_links.shape[0]
print(f'Total number of messages with links: {total_links}')

#total number of users
total_users = df['sender_id'].nunique()
print(f'Total number of unique users: {total_users}')

#total number of users sending messages with links
users_with_links = df_links['sender_id'].nunique()
print(f'Number of unique users who sent messages with links: {users_with_links}')

# Average number of messages with links per user
avg_links_per_user = total_links / total_users
print(f'Average number of messages with links per user: {avg_links_per_user:.2f}')

# Average number of messages with links per user who sent at least one message with a link
avg_links_per_link_sender = total_links / users_with_links
print(f'Average number of messages with links per user who sent at least one message with a link: {avg_links_per_link_sender:.2f}')
```

## Ideology check of users

```python
# retrieve top 10 users
top_10_users=df_links['sender_id'].value_counts().head(10).index.tolist()
print("Top 10 most active users sending messages with links:")
print(top_10_users)
# Ideological orientation of top 10 users sending messages with links
top_10_users_info = df_users[df_users['sender_id'].isin(top_10_users)][['sender_id', 'ideology_score', 'credibility_score']]
top_10_users_info = top_10_users_info.set_index('sender_id').loc[top_10_users].reset_index()

print("Ideological orientation of top 10 users sending messages with links:")   
print(top_10_users_info)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D

# Build plotting table from your existing variables
# top_link_senders: Series indexed by sender_id with count of link messages
# top_10_users_info: DataFrame with sender_id, ideology_score, credibility_score
plot_df = (
    top_link_senders.rename_axis("sender_id")
    .reset_index(name="link_messages")
    .merge(top_10_users_info, on="sender_id", how="left")
)

# Keep ranking order from activity
plot_df["rank"] = np.arange(1, len(plot_df) + 1)
plot_df["label"] = "U" + plot_df["rank"].astype(str)

# Classify ideology into 3 groups using a small neutral band around zero
neutral_threshold = 5
plot_df["ideology_group"] = np.where(
    plot_df["ideology_score"] < -neutral_threshold,
    "Mostly left-wing",
    np.where(
        plot_df["ideology_score"] > neutral_threshold,
        "Mostly right-wing",
        "Neutral"
    )
)

# Inverse mapping requested: left=red, right=blue
group_order = ["Mostly left-wing", "Neutral", "Mostly right-wing"]
group_colors = {
    "Mostly left-wing": "#d73027",
    "Neutral": "#9e9e9e",
    "Mostly right-wing": "#4477ba"
}

# Keep original bubble sizing in the plot
size_scale = 25

# Fixed legend references for bubble sizes (messages)
size_legend_values = [40, 90, 150]

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6), gridspec_kw={"width_ratios": [1.0, 1.4]})

# ---- Left: activity ranking with ideology colors ----
bar_df = plot_df.sort_values("link_messages", ascending=True).copy()
bar_colors = bar_df["ideology_group"].map(group_colors).fillna("#9e9e9e")
axes[0].barh(bar_df["label"], bar_df["link_messages"], color=bar_colors, alpha=0.9)
axes[0].set_title("Top users by messages with links")
axes[0].set_xlabel("Messages with links")
axes[0].set_ylabel("User (rank)")

# ---- Right: ideology vs credibility bubble plot ----
for group_name in group_order:
    group_data = plot_df[plot_df["ideology_group"] == group_name]
    if group_data.empty:
        continue
    axes[1].scatter(
        group_data["ideology_score"],
        group_data["credibility_score"],
        s=group_data["link_messages"] * size_scale,
        color=group_colors[group_name],
        alpha=0.85,
        edgecolor="black",
        linewidth=0.6
    )

# Quadrant lines at real zero references
axes[1].axhline(0, color="gray", linestyle="--", linewidth=1)
axes[1].axvline(0, color="gray", linestyle="--", linewidth=1)

# Label each point with rank
for _, r in plot_df.iterrows():
    axes[1].annotate(
        r["label"],
        (r["ideology_score"], r["credibility_score"]),
        xytext=(6, 5),
        textcoords="offset points",
        fontsize=9,
        weight="bold"
    )

# Color legend (fixed order)
color_handles = [
    Line2D([0], [0], marker="o", color="w", label=group,
           markerfacecolor=group_colors[group], markeredgecolor="black", markersize=9)
    for group in group_order
    if (plot_df["ideology_group"] == group).any()
 ]

# Size legend with fixed references
size_handles = [
    plt.scatter([], [], s=v * size_scale, c="#bdbdbd", alpha=0.85, edgecolors="black", linewidth=0.6)
    for v in size_legend_values
]
size_labels = [f"{v} messages" for v in size_legend_values]

axes[1].set_title("Ideology vs credibility (size = activity, color = ideology group)")
axes[1].set_xlabel("Ideology score")
axes[1].set_ylabel("Credibility score")

legend_colors = axes[1].legend(
    handles=color_handles,
    title="Ideology group",
    loc="upper left",
    bbox_to_anchor=(1.02, 1.0),
    borderaxespad=0.0,
    frameon=True
)
axes[1].add_artist(legend_colors)

axes[1].legend(
    handles=size_handles,
    labels=size_labels,
    title="Bubble size",
    loc="upper left",
    bbox_to_anchor=(1.02, 0.52),
    borderaxespad=0.0,
    frameon=True,
    scatterpoints=1
)

# Note for classification threshold
fig.text(
    0.64, 0.02,
    f"Neutral definition: |ideology score| <= {neutral_threshold}",
    ha="center",
    va="bottom",
    fontsize=9,
    color="dimgray"
 )

fig.subplots_adjust(right=0.82, bottom=0.12)
plt.tight_layout()
plt.show()
```
